```python
from rapidfireai.automl import RFGridSearch, RFModelConfig

cfg = RFModelConfig(model_name="Qwen/Qwen2.5-0.5B-Instruct", num_gpus=2)
config_leaf = RFGridSearch(configs=[cfg], trainer_type="SFT")._get_runs_fit()[0]

default_req_workers = 1
required_workers = config_leaf.get("num_gpus", default_req_workers)
use_fsdp = (
    "training_args" in config_leaf
    and "fsdp_config" in config_leaf["training_args"]
)

print(required_workers)
print(use_fsdp)
```
